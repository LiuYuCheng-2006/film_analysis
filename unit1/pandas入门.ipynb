```python
import pandas as pd
df=pd.DataFrame([{"姓名":"小王", "语文":90, "数学":80, "英语":70},
              {"姓名":"小王2", "语文":78, "数学":83, "英语":72},
              {"姓名":"小王3", "语文":30, "数学":50, "英语":60},
              {"姓名":"小王4", "语文":40, "数学":65, "英语":20},
              ])
print(f"语文最高分:{df["语文"].max()},语文最低分：{df["语文"].min()},语文平均分：{df["语文"].mean()}")
print(f"数学最高分:{df["数学"].max()},数学最低分:{df["数学"].min()},数学平均分:{df["数学"].mean()}")
print(f"英语最高分:{df["英语"].max()},英语最低分:{df["英语"].min()},英语平均分：{df["英语"].mean()}")
```
